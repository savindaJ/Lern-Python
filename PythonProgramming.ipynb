# Dictionary operations

```python
# Create a dictionary
my_dict = {"name": "Alice", "age": 25, "city": "New York"}

# Access a value using a key
print(my_dict["name"])  # Output: Alice

# Add a new key-value pair
my_dict["email"] = "alice@example.com"
print(my_dict)

# Update an existing value
my_dict["age"] = 26
print(my_dict)

# Remove a key-value pair
del my_dict["city"]
print(my_dict)

# Iterate through keys
for key in my_dict:
    print(key, my_dict[key])

# Iterate through key-value pairs
for key, value in my_dict.items():
    print(key, value)

# Check if a key exists
if "name" in my_dict:
    print("Name is present in the dictionary")

# Get the value of a key with a default value if the key does not exist
print(my_dict.get("phone", "Not Available"))
```

```python
# Pop a value from the dictionary using a key
popped_value = my_dict.pop("name", "Key not found")
print(popped_value)
print(my_dict)
```

```python
# Arithmetic Operators
a = 10
b = 5

print("Addition:", a + b)        # Output: 15
print("Subtraction:", a - b)     # Output: 5
print("Multiplication:", a * b)  # Output: 50
print("Division:", a / b)        # Output: 2.0
print("Modulus:", a % b)         # Output: 0
print("Exponentiation:", a ** b) # Output: 100000
print("Floor Division:", a // b) # Output: 2

# Comparison Operators
print("Equal:", a == b)          # Output: False
print("Not Equal:", a != b)      # Output: True
print("Greater than:", a > b)    # Output: True
print("Less than:", a < b)       # Output: False
print("Greater than or equal to:", a >= b) # Output: True
print("Less than or equal to:", a <= b)    # Output: False

# Logical Operators
x = True
y = False

print("AND:", x and y)           # Output: False
print("OR:", x or y)             # Output: True
print("NOT:", not x)             # Output: False

# Bitwise Operators
print("Bitwise AND:", a & b)     # Output: 0
print("Bitwise OR:", a | b)      # Output: 15
print("Bitwise XOR:", a ^ b)     # Output: 15
print("Bitwise NOT:", ~a)        # Output: -11
print("Bitwise Left Shift:", a << 1) # Output: 20
print("Bitwise Right Shift:", a >> 1) # Output: 5

# Assignment Operators
c = a + b
print("Assignment:", c)          # Output: 15

c += a
print("Add and Assign:", c)      # Output: 25

c -= a
print("Subtract and Assign:", c) # Output: 15

c *= a
print("Multiply and Assign:", c) # Output: 150

c /= a
print("Divide and Assign:", c)   # Output: 15.0

c %= a
print("Modulus and Assign:", c)  # Output: 5.0

c **= a
print("Exponent and Assign:", c) # Output: 3125.0

c //= a
print("Floor Divide and Assign:", c) # Output: 312.0

# Identity Operators
print("Is:", a is b)             # Output: False
print("Is Not:", a is not b)     # Output: True

# Membership Operators
list1 = [1, 2, 3, 4, 5]
print("In:", a in list1)         # Output: False
print("Not In:", a not in list1) # Output: True
```

```python
num1 = 10
num2 = 20
operator = "+"

def calclate(num1 , num2 , operator):
    if operator == "+":
        return num1 + num2
    elif operator == "-":
        return num1 - num2
    elif operator == "*":
        return num1 * num2
    elif operator == "/":
        return num1 / num2
    else:
        return "Invalid Operator"
    
print(calclate(num1 , num2 , operator))
```

# Control Flows

```python
# If-Else Statement
age = 18
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")

# Elif Statement
score = 85
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")

# For Loop
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

# While Loop
count = 0
while count < 5:
    print(count)
    count += 1

# Break Statement
for i in range(10):
    if i == 5:
        break
    print(i)
else:
    print("for loop was stoped !")

# Continue Statement
for i in range(10):
    if i == 5:
        continue
    print(i)

# Pass Statement
for i in range(10):
    if i == 5:
        pass
    print(i)
```
