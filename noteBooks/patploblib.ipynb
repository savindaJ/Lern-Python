# predicting Houses Price

## problem Statement

You are provided with a dataset containing information about houses, including features like the size of the house (in square feet) and the number of bedrooms. The goal is to predict the price of a house based on these features using linear regression.


| Size (sqft) | Bedrooms | Price ($) |
|-------------|----------|-----------|
| 2104        | 3        | 399900    |
| 1600        | 3        | 329900    |
| 2400        | 3        | 369000    |
| 1416        | 2        | 232000    |
| 3000        | 4        | 539900    |

# Load Data set

```python
%pip install pandas
```

```python
import pandas as pd
data = {
    "Size (sqft)": [2104, 1600, 2400, 1416, 3000],
    "Bedrooms": [3, 3, 3, 2, 4],
    "Price ($)": [399900, 329900, 369000, 232000, 539900]
}
df = pd.DataFrame(data)
print(df)
```

# visualize Data

```python
import matplotlib.pyplot as plt

plt.scatter(df['Size (sqft)'], df['Price ($)'], color='blue' , label='Size')
plt.xlabel('Size (sqft)')
plt.ylabel('Price ($)')
plt.title('House price prediction')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df['Bedrooms'], df['Price ($)'], color='blue' , label='Size')
plt.xlabel('Size (sqft)')
plt.ylabel('Price ($)')
plt.title('House price prediction')
plt.legend()
plt.show()
```

# Prepere Data set

```python
x = df[['Size (sqft)', 'Bedrooms']].values
y = df['Price ($)'].values
```

# Define Aad Train

```python
%pip install scikit-learn
```

```python
from sklearn.linear_model import LinearRegression

modal = LinearRegression()
modal.fit(x, y)
```

```python
theta_0 = modal.intercept_ # Price
theta_1 = modal.coef_[0] # Size (sqft)
theta_2 = modal.coef_[1] # Bedrooms

print(f"Price = {theta_0}")
print(f"Size = {theta_1}")
print(f"Bedrooms = {theta_2}")
```

# get Prediction

```python
predictions = modal.predict(x) # Size (sqft), Bedrooms using the modal we trained above , Price prediction for a house with 2500 sqft and 3 bedrooms
print(f"Prediction: {predictions[0]}")
```

# Modal Evaluation

```python
comparison = pd.DataFrame({'Actual': y, 'Predicted': predictions , 'Difference': y - predictions})
print(comparison)
```

# Drug dosage nd blood presure

```python
drucg_and_bllod_pressure = {
    "Drug": [10 , 15 , 20 , 25 , 30 , 35 , 40 , 45 , 50 , 55],
    "Blood Pressure": [120, 125, 132, 140, 145, 150, 158, 165, 170, 178]
}

df = pd.DataFrame(drucg_and_bllod_pressure)
print(df)

import matplotlib.pyplot as plt

plt.scatter(df['Drug'], df['Blood Pressure'], color='blue' , label='Size')
plt.xlabel('Drug')
plt.ylabel('Blood Pressure')
plt.title('Drug and Blood Pressure')
plt.legend()
plt.show()

x = df[['Drug']].values
y = df['Blood Pressure'].values

modal = LinearRegression()

modal.fit(x, y)

theta_0 = modal.intercept_ # Blood Pressure
theta_1 = modal.coef_[0] # Drug

print(f"Blood Pressure = {theta_0}")
print(f"Drug = {theta_1}")

predictions = modal.predict([[8]])
print(f"Prediction Blood Presure: {predictions[0]}")

comparison = pd.DataFrame({'Actual': y, 'Predicted': modal.predict(x) , 'Difference': y - modal.predict(x)})
print(comparison)

```
