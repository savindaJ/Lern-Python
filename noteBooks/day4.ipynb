```python
def is_sorted(numbers):
    numbers = numbers.split()
    return sorted(numbers)

numbers = str(input('Enter Numbers: '))

try:
    for number in numbers.split():
        int(number)
except ValueError:
    raise ValueError('Please enter only numbers')

print('Sorted Numbers:', is_sorted(numbers))
```

# Array Operators

## arithmatic Operators

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

sum = a + b
print(sum)

diff = a - b
print(diff)

prod = a * b
print(prod)

div = a / b
print(div)

dot = np.dot(a, b)
print(dot)
```

# Metriks Multiplicaton

```python
a = np.array([[0 , 1] , [2 , 3]])
b = np.array([[4 , 5] , [6 , 7]])

dot = np.dot(a, b)
print(dot)

print('Transpose')

a = np.array([[0 , 1] , [2 , 3]])
b = np.array([[4 , 5] , [6 , 7]])

transpose = a.T + b.T

print(transpose)
```
